### Primer experimento: CNN inicial

Vamos a partir de una CNN sencilla que servirá de referencia para todos los experimentos posteriores. Se va a validar una base clara y ver qué cambia cuando se introduzca modificaciones una a una.

La red tiene dos capas convolucionales, una de 32 filtros y otra de 64, ambos con kernels de 3 × 3 y activación ReLU. Tras cada convolución aplicaremos MaxPooling para reducir el tamaño de los mapas de características. Podriamo pensar en aumentar los filtros en el segundo bloque porque las capas más profundas combinan patrones simples en otros más complejos y eligir kernels pequeños porque los defectos buscados son detalles de pocos píxeles. Después aplanamos la salida y se añadiria una capa densa de 64 neuronas con ReLU antes de la capa final.

Entrada y salida: Las imágenes entran como tensores de 128 × 128 × 3, y la primera capa del modelo las normaliza dividiendo entre 255. Como el problema tiene solo dos clases, la salida es una neurona con activación sigmoide que da la probabilidad de que la pieza sea correcta (la clase ok_front está codificada como 1). La función de pérdida es binary_crossentropy, la adecuada para clasificación binaria con esta salida.

Entrenamiento. Utilizamos optimizador Adam con su tasa de aprendizaje por defecto (0,001), lotes de 32 imágenes y 10 épocas. Se fijan estos valores de forma idéntica para todos los experimentos de CNN para que las comparaciones sean justas y los analizaremos según lo que muestren las curvas.

Tamaño del modelo: Antes de entrenar comprobamos en el resumen del modelo que tiene unos 3,7 millones de parámetros, y que casi todos (más del 99%) están en la capa densa que sigue al aplanado. Las convolucionales apenas suman unos 19.000. Esto puede hacer anticipar un riesgo de sobreajuste, porque la parte del modelo con más capacidad es la que menos explota la estructura de la imagen. Será una de las razones para probar después Dropout y regularización L2.

Vamos a partir de un primer experimento sin incorporar Dropout, Batch Normalization, L2 ni Data Augmentation para tener una referencia sin estas técnicas y poder medir el efecto que tenga cada una cuando se añada.

Miraremos sobre todo las distancias entre las curvas de entrenamiento y validación, si la pérdida de validación deja de bajar o empieza a subir o si la validación oscila entre épocas. Esos signos pueden indicar si el modelo sobreajusta o es inestable y orientarán qué técnica probar.

```python

from tensorflow.keras import layers, models

# Voy a crear mi primera red CNN con dos capas
# de convolución para comenzar con un modelo sencillo.

modelo_cnn = models.Sequential([

    # Normalizamos los valores de los píxeles.
    layers.Input(shape=(128, 128, 3)),
    layers.Rescaling(1./255),

    # Primera capa de convolución.
    layers.Conv2D(32, (3, 3), activation='relu'),
    layers.MaxPooling2D((2, 2)),

    # Segunda capa de convolución.
    layers.Conv2D(64, (3, 3), activation='relu'),
    layers.MaxPooling2D((2, 2)),

    # Convertimos los datos para realizar la clasificación.
    layers.Flatten(),
    layers.Dense(64, activation='relu'),

    # La salida indica si la pieza es correcta o defectuosa.
    layers.Dense(1, activation='sigmoid')
])

# Revisamos las capas y parámetros del modelo.
modelo_cnn.summary()
```

```python

# Configuramos el modelo para comenzar el entrenamiento. Utilizamos binary_crossentropy porque tenemos dos clases.

modelo_cnn.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

print("Modelo preparado para entrenar.")
```

```python

# Entrenamos nuestra primera CNN durante 10 épocas.
# Guardamos los resultados para revisar las gráficas después.

resultado_cnn = modelo_cnn.fit(
    imagenes_train,
    validation_data=imagenes_val,
    epochs=10
)
```

```python

import matplotlib.pyplot as plt

# Revisamos cómo ha cambiado el accuracy y el error
# durante las 10 épocas de nuestro primer modelo.

plt.figure(figsize=(12, 4))

# Gráfica del accuracy.
plt.subplot(1, 2, 1)
plt.plot(resultado_cnn.history['accuracy'], label='Entrenamiento')
plt.plot(resultado_cnn.history['val_accuracy'], label='Validación')
plt.title('Accuracy de mi primera CNN')
plt.xlabel('Épocas')
plt.ylabel('Accuracy')
plt.legend()

# Gráfica del error.
plt.subplot(1, 2, 2)
plt.plot(resultado_cnn.history['loss'], label='Entrenamiento')
plt.plot(resultado_cnn.history['val_loss'], label='Validación')
plt.title('Loss de mi primera CNN')
plt.xlabel('Épocas')
plt.ylabel('Loss')
plt.legend()

plt.tight_layout()
plt.show()
```

### Resultados del primer experimento

La primera CNN ha aprendido bastante rápido. En entrenamiento pasó de un accuracy del 70,96% en la primera época al 99,76% en la décima. En validación terminó con un 98,94% que parece un buen punto de partida para una red tan sencilla. La pérdida de validación bajó de 0,3561 a 0,0454.

En la séptima época la validación cayó al 94,57% (con una pérdida de 0,1500) cuando venía de un 98,42%, y se recuperó en la octava. Estas oscilaciones indican que el entrenamiento es algo inestable con esta configuración, aunque en las últimas épocas las curvas se estabilizan.

Aunque la mejor validación coincide con la última época, la inestabilidad de la época 7  impide asegurar que más épocas den necesariamente un mejor resultado. Establecemos esta CNN como referencia y para probar las técnicas de regularización.

### Segundo experimento: CNN con Dropout

La CNN inicial consiguió un accuracy alto, pero ahora veremos si se puede controlar mejor la diferencia entre entrenamiento y validación. Para esta prueba introduciremos Dropout con una tasa de 0.3.

Durante el entrenamiento, Dropout deja inactivas de forma aleatoria algunas neuronas de la capa donde se aplica. Con ello buscamos que la red no dependa tanto de unas conexiones concretas.

Mantendremos el resto de la estructura y las diez épocas. Así podremos comparar este cambio con la primera CNN sin mezclar varias técnicas a la vez.

```python

from tensorflow.keras import layers, models

# Creamos otra CNN parecida a la primera,
# pero esta vez agregamos Dropout.

modelo_dropout = models.Sequential([

    layers.Input(shape=(128, 128, 3)),
    layers.Rescaling(1./255),

    layers.Conv2D(32, (3, 3), activation='relu'),
    layers.MaxPooling2D((2, 2)),

    layers.Conv2D(64, (3, 3), activation='relu'),
    layers.MaxPooling2D((2, 2)),

    layers.Flatten(),
    layers.Dense(64, activation='relu'),

    # Agregamos Dropout para intentar reducir el sobreajuste.
    layers.Dropout(0.3),

    layers.Dense(1, activation='sigmoid')
])

modelo_dropout.summary()
```

```python

# Entrenamos el segundo modelo con las mismas
# condiciones del primero para poder compararlos.

modelo_dropout.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

resultado_dropout = modelo_dropout.fit(
    imagenes_train,
    validation_data=imagenes_val,
    epochs=10
)
```

```python

# Revisamos cómo se comportó nuestra CNN con Dropout.

plt.figure(figsize=(12, 4))

plt.subplot(1, 2, 1)
plt.plot(resultado_dropout.history['accuracy'], label='Entrenamiento')
plt.plot(resultado_dropout.history['val_accuracy'], label='Validación')
plt.title('Accuracy CNN con Dropout')
plt.xlabel('Épocas')
plt.ylabel('Accuracy')
plt.legend()

plt.subplot(1, 2, 2)
plt.plot(resultado_dropout.history['loss'], label='Entrenamiento')
plt.plot(resultado_dropout.history['val_loss'], label='Validación')
plt.title('Loss CNN con Dropout')
plt.xlabel('Épocas')
plt.ylabel('Loss')
plt.legend()

plt.tight_layout()
plt.show()
```

### Resultados del segundo experimento

Con Dropout hemos obtenido un 98,79% de accuracy en entrenamiento y un 98,87% en validación al finalizar. En accuracy de validación no hay ventaja sobre la CNN inicial que terminó en 98,94%. La pérdida de validación sí fue algo menor, 0,0412 frente a 0,0454.

Lo que sí cambia es el comportamiento durante el entrenamiento, la CNN inicial tuvo una caída de la validación en la época 7 (94,57%), mientras que con Dropout las curvas son más regulares a partir de la época 4 y el accuracy de validación se mantiene entre 98,2% y 98,9%. Además la distancia entre entrenamiento y validación casi desaparece.

En esta prueba Dropout parece haber estabilizado el entrenamiento pero no ha mejorado el accuracy final y las diferencias son demasiado pequeñas para distinguirlas del ruido de una sola ejecución.

### Tercer experimento: CNN con Batch Normalization

En esta ocasión colocaremos Batch Normalization después de las dos capas convolucionales para observar cómo influye en el entrenamiento.

No combinaremos ambas técnicas, porque entonces sería más difícil saber a cuál atribuir los cambios. Mantenemos las dos convoluciones, el tamaño de las imágenes y las diez épocas.

Miraremos en las gráficas si la validación se comporta de una manera más regular o si aparecen variaciones importantes.

```python

from tensorflow.keras import layers, models

# Creamos una nueva CNN para probar Batch Normalization.
# Mantenemos las dos capas convolucionales del primer modelo.

modelo_batch = models.Sequential([

    layers.Input(shape=(128, 128, 3)),
    layers.Rescaling(1./255),

    # Primera convolución con Batch Normalization.
    layers.Conv2D(32, (3, 3), activation='relu'),
    layers.BatchNormalization(),
    layers.MaxPooling2D((2, 2)),

    # Segunda convolución con Batch Normalization.
    layers.Conv2D(64, (3, 3), activation='relu'),
    layers.BatchNormalization(),
    layers.MaxPooling2D((2, 2)),

    layers.Flatten(),
    layers.Dense(64, activation='relu'),

    layers.Dense(1, activation='sigmoid')
])

modelo_batch.summary()
```

```python

# Entrenamos el tercer modelo durante 10 épocas
# para poder comparar con los anteriores.

modelo_batch.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

resultado_batch = modelo_batch.fit(
    imagenes_train,
    validation_data=imagenes_val,
    epochs=10
)
```

### Cuarto experimento: CNN con Data Augmentation

Hasta ahora hemos cambiado el funcionamiento interno de la CNN pero ahora con Data Augmentation vamos a probar la idea de introducir pequeñas variaciones en las imágenes mientras se entrena el modelo.

Utilizaremos cambios moderados de orientación y tamaño, además de giros horizontales. La intención es que la red no dependa de que todas las fotografías estén colocadas exactamente igual.

En piezas industriales hay que tener cuidado con esto. Si modificamos demasiado una imagen podría alterar justo el detalle que permite reconocer un defecto. Por eso elegimos transformaciones pequeñas manteniendo la estructura de la CNN inicial.

```python

from tensorflow.keras import layers, models

# Aplicamos pequeñas variaciones a las imágenes
# para probar si ayudan al entrenamiento.

cambios_imagenes = models.Sequential([
    layers.RandomFlip("horizontal"),
    layers.RandomRotation(0.05),
    layers.RandomZoom(0.05)
])

# Creamos otra CNN utilizando la misma estructura inicial.
modelo_aumento = models.Sequential([

    layers.Input(shape=(128, 128, 3)),

    # Aplicamos los cambios durante el entrenamiento.
    cambios_imagenes,

    layers.Rescaling(1./255),

    layers.Conv2D(32, (3, 3), activation='relu'),
    layers.MaxPooling2D((2, 2)),

    layers.Conv2D(64, (3, 3), activation='relu'),
    layers.MaxPooling2D((2, 2)),

    layers.Flatten(),
    layers.Dense(64, activation='relu'),

    layers.Dense(1, activation='sigmoid')
])

modelo_aumento.summary()
```

```python

# Entrenamos el cuarto modelo durante 10 épocas para comparar con los experimentos anteriores.

modelo_aumento.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

resultado_aumento = modelo_aumento.fit(
    imagenes_train,
    validation_data=imagenes_val,
    epochs=10
)
```

```python

# Revisamos los resultados con Data Augmentation.

plt.figure(figsize=(12, 4))

plt.subplot(1, 2, 1)
plt.plot(resultado_aumento.history['accuracy'], label='Entrenamiento')
plt.plot(resultado_aumento.history['val_accuracy'], label='Validación')
plt.title('Accuracy CNN con Data Augmentation')
plt.xlabel('Épocas')
plt.ylabel('Accuracy')
plt.legend()

plt.subplot(1, 2, 2)
plt.plot(resultado_aumento.history['loss'], label='Entrenamiento')
plt.plot(resultado_aumento.history['val_loss'], label='Validación')
plt.title('Loss CNN con Data Augmentation')
plt.xlabel('Épocas')
plt.ylabel('Loss')
plt.legend()

plt.tight_layout()
plt.show()
```

### Resultados del cuarto experimento

La CNN con Data Augmentation terminó con un accuracy del 98,10% en entrenamiento y del 97,21% en validación. La pérdida de validación acabó en 0,0975. Estos valores quedan por debajo de los de la CNN inicial (98,94% y 0,0454) y de Dropout (98,87% y 0,0412), de modo que en esta prueba no observamos ninguna mejora sino un resultado algo peor.

Las curvas muestran un aprendizaje más lento y más irregular. La validación cayó hasta el 70,66% y el 68,40% en las épocas 3 y 4, bajó de nuevo al 90,05% en la séptima y solo se estabilizó en las últimas épocas. Además el accuracy de entrenamiento seguía aumentando en la décima época (de 96,55% a 98,10%), por lo que el modelo probablemente no había terminado de converger. Esto en parte es esperable porque las imágenes se modifican en cada época y la red tarda más en ajustarse. Conviene recordar también que el accuracy de entrenamiento se calcula sobre imágenes transformadas, que son más difíciles que las originales.

Con estas diez épocas no podemos concluir que Data Augmentation sea perjudicial solo que con las transformaciones utilizadas (giro horizontal, rotación y zoom del 5%) y este número de épocas no ha mejorado a la CNN inicial. En la evaluación final comprobaremos si entrenar con imágenes modificadas ayuda a reconocer mejor los defectos con fotografías que el modelo no ha visto.

### Quinto experimento: CNN con regularización L2

Para completar las pruebas de regularización vamos a utilizar L2. A diferencia de Data Augmentation, aquí no cambiaremos las fotografías ni a desactivar neuronas: la idea es penalizar los pesos grandes de la red durante el entrenamiento.

Aplicaremos L2 con un valor de 0.001 en las capas convolucionales y en la capa densa de 64 neuronas. La estructura general seguirá siendo la de la primera CNN y diez épocas.

Revisaremos si esta restricción ayuda a que el modelo generalice mejor. También tendremos presente que el loss ahora incluye la penalización L2 por lo que no sería correcto comparar directamente ese valor con el de una CNN sin regularización.

```python

from tensorflow.keras import layers, models, regularizers

# Creamos otra CNN parecida a la primera,
# pero esta vez aplicamos regularización L2.

modelo_l2 = models.Sequential([

    layers.Input(shape=(128, 128, 3)),
    layers.Rescaling(1./255),

    layers.Conv2D(
        32, (3, 3),
        activation='relu',
        kernel_regularizer=regularizers.l2(0.001)
    ),
    layers.MaxPooling2D((2, 2)),

    layers.Conv2D(
        64, (3, 3),
        activation='relu',
        kernel_regularizer=regularizers.l2(0.001)
    ),
    layers.MaxPooling2D((2, 2)),

    layers.Flatten(),

    layers.Dense(
        64,
        activation='relu',
        kernel_regularizer=regularizers.l2(0.001)
    ),

    layers.Dense(1, activation='sigmoid')
])

modelo_l2.summary()
```

```python

# Entrenamos nuestra CNN con regularización L2.
# Mantenemos las 10 épocas de los otros experimentos.

modelo_l2.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

resultado_l2 = modelo_l2.fit(
    imagenes_train,
    validation_data=imagenes_val,
    epochs=10
)
```

```python

# Revisamos el accuracy y el loss de nuestra CNN con L2.

plt.figure(figsize=(12, 4))

plt.subplot(1, 2, 1)
plt.plot(resultado_l2.history['accuracy'], label='Entrenamiento')
plt.plot(resultado_l2.history['val_accuracy'], label='Validación')
plt.title('Accuracy CNN con L2')
plt.xlabel('Épocas')
plt.ylabel('Accuracy')
plt.legend()

plt.subplot(1, 2, 2)
plt.plot(resultado_l2.history['loss'], label='Entrenamiento')
plt.plot(resultado_l2.history['val_loss'], label='Validación')
plt.title('Loss CNN con L2')
plt.xlabel('Épocas')
plt.ylabel('Loss')
plt.legend()

plt.tight_layout()
plt.show()
```

### Resultados del quinto experimento

En L2 encontramos una situación que no se aprecia mirando solamente la última época. La validación llegó a un accuracy del 97,96% en las épocas 6 y 7, pero después empezó a bajar (97,44%, 94,95%) y terminó en 93,51%. El entrenamiento finalizó en 97,23%, tras haber llegado a 98,55% en la época 9.

La pérdida de validación tuvo su mínimo en la época 6 (0,1559) y subió hasta 0,2469 en la última. Esta pérdida incluye la penalización L2, por lo que solo la comparamos con las épocas anteriores de este mismo modelo, no con la de los experimentos sin regularización. Como la pérdida y el accuracy de entrenamiento también empeoraron en la última época, no lo interpretamos como un sobreajuste claro, sino más bien como una señal de que la configuración no está funcionando de forma estable.

No diríamos que la técnica es mala por este único experimento pero si se podría afirmar que con el valor 0,001 y las diez épocas utilizadas no hemos conseguido una mejora frente a la CNN inicial: ni el mejor valor de validación de L2 (97,96%) ni el final (93,51%) superan el 98,94% de la CNN inicial.

### Sexto experimento: Transfer Learning con MobileNetV2

Después de las cinco pruebas con CNN cambiaremos el enfoque, en lugar de construir toda la red desde cero, utilizaremos MobileNetV2 con los pesos que ya aprendió en ImageNet.

Quitaremos la clasificación original y añadiremos una salida adaptada a nuestras dos categorías. Por ahora dejaremos congeladas las capas preentrenadas de manera que solo se entrenen las capas nuevas. Con esto lo que se espera comprobar es si aprovechar las características aprendidas por otro modelo ofrece alguna ventaja al trabajar con estas piezas industriales.

```python

import tensorflow as tf
from tensorflow.keras import layers, models

# Cargamos MobileNetV2 con los pesos de ImageNet.
# No incluimos su clasificación original porque
# nosotros solamente tenemos dos categorías.

modelo_base = tf.keras.applications.MobileNetV2(
    input_shape=(128, 128, 3),
    include_top=False,
    weights='imagenet'
)

# Dejamos congeladas las capas originales.
modelo_base.trainable = False

# Creamos nuestro modelo para clasificar las piezas.
modelo_mobilenet = models.Sequential([
    layers.Input(shape=(128, 128, 3)),

    # Preparamos los valores como necesita MobileNetV2.
    layers.Rescaling(1./127.5, offset=-1),

    modelo_base,

    layers.GlobalAveragePooling2D(),
    layers.Dense(64, activation='relu'),
    layers.Dense(1, activation='sigmoid')
])

modelo_mobilenet.summary()
```

```python

# Entrenamos solamente las capas nuevas de clasificación.

modelo_mobilenet.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
    loss='binary_crossentropy',
    metrics=['accuracy']
)

resultado_mobilenet = modelo_mobilenet.fit(
    imagenes_train,
    validation_data=imagenes_val,
    epochs=10
)
```

```python

# Revisamos el accuracy y el loss de MobileNetV2.

plt.figure(figsize=(12, 4))

plt.subplot(1, 2, 1)
plt.plot(resultado_mobilenet.history['accuracy'],
         label='Entrenamiento')
plt.plot(resultado_mobilenet.history['val_accuracy'],
         label='Validación')
plt.title('Accuracy MobileNetV2')
plt.xlabel('Épocas')
plt.ylabel('Accuracy')
plt.legend()

plt.subplot(1, 2, 2)
plt.plot(resultado_mobilenet.history['loss'],
         label='Entrenamiento')
plt.plot(resultado_mobilenet.history['val_loss'],
         label='Validación')
plt.title('Loss MobileNetV2')
plt.xlabel('Épocas')
plt.ylabel('Loss')
plt.legend()

plt.tight_layout()
plt.show()
```

### Resultados del sexto experimento

Con MobileNetV2 y las capas originales congeladas el entrenamiento avanzó rápidamente. El accuracy final fue del 99,55% en entrenamiento y del 98,64% en validación con una pérdida de validación de 0,0409. El mejor valor de validación fue 99,17%, en la época 9.

En las gráficas se aprecia que el modelo ya parte de valores altos (97,66% de validación en la primera época) y que las variaciones de validación son relativamente pequeñas. Esto tiene sentido porque no estamos aprendiendo todas las características desde cero sino aprovechando el preentrenamiento y entrenando solo las capas nuevas.

Frente a la CNN inicial, MobileNetV2 converge antes y es más estable, aunque el accuracy final de validación es parecido (98,64% frente a 98,94%). Es decir, el modelo preentrenado no supera claramente a la red sencilla en validación por lo que tocaría ver si esa igualdad se mantiene con las imágenes de prueba.

### Séptimo experimento: Fine-tuning con MobileNetV2

Ahora vamos a permitir que parte de MobileNetV2 se ajuste a las imágenes. Trabajaremos con las últimas 20 capas, manteniendo congeladas las anteriores y las capas Batch Normalization.

Antes de hacer ese cambio guardaremos la versión con las capas originales congeladas. Así podremos recuperar ambos modelos y compararlos al final, sin confundir sus resultados.

Utilizaremos una tasa de aprendizaje más pequeña para evitar modificar demasiado los pesos preentrenados. Lo que nos interesa validar es si este ajuste ayuda a reconocer mejor los defectos de las piezas.

```python

# Revisamos cómo se comportó MobileNetV2
# después de aplicar fine-tuning.

plt.figure(figsize=(12, 4))

plt.subplot(1, 2, 1)
plt.plot(resultado_fine.history['accuracy'],
         label='Entrenamiento')
plt.plot(resultado_fine.history['val_accuracy'],
         label='Validación')
plt.title('Accuracy MobileNetV2 Fine-tuning')
plt.xlabel('Épocas')
plt.ylabel('Accuracy')
plt.legend()

plt.subplot(1, 2, 2)
plt.plot(resultado_fine.history['loss'],
         label='Entrenamiento')
plt.plot(resultado_fine.history['val_loss'],
         label='Validación')
plt.title('Loss MobileNetV2 Fine-tuning')
plt.xlabel('Épocas')
plt.ylabel('Loss')
plt.legend()

plt.tight_layout()
plt.show()
```

El fine-tuning de MobileNetV2 terminó con un accuracy del 99,81% en entrenamiento y del 99,25% en validación con una pérdida de validación de 0,0456. Adicionalmente, esta versión acumula veinte épocas de entrenamiento frente a diez, por lo que cualquier mejora podría deberse a ese entrenamiento adicional y no al ajuste de capas, todo esto hace que se puedan sacar conclusiones sobre cuánto ha ayudado el fine-tuning.

### Octavo experimento: Transfer Learning con ResNet50

La segunda arquitectura preentrenada que vamos utilizar es ResNet50. Igual que con MobileNetV2 comenzaremos cargando sus pesos de ImageNet y manteniendo congelada la red original.

La diferencia es que ResNet50 tiene una arquitectura más grande. No esperamos que por eso sea necesariamente mejor, precisamente queremos comprobar si esa mayor complejidad se traduce en mejores clasificaciones en nuestro dataset Añadiremos las capas finales para nuestras dos clases y mantendremos las diez épocas de entrenamiento.

```python

from tensorflow.keras import layers, models
import tensorflow as tf

# Cargamos ResNet50 con los pesos de ImageNet.
# No utilizamos su clasificación original.

base_resnet = tf.keras.applications.ResNet50(
    input_shape=(128, 128, 3),
    include_top=False,
    weights='imagenet'
)

# Congelamos las capas originales.
base_resnet.trainable = False

# Creamos nuestro modelo para clasificar las piezas.
modelo_resnet = models.Sequential([

    layers.Input(shape=(128, 128, 3)),

    # Preparamos las imágenes como necesita ResNet50.
    layers.Lambda(
        tf.keras.applications.resnet50.preprocess_input
    ),

    base_resnet,

    layers.GlobalAveragePooling2D(),
    layers.Dense(64, activation='relu'),
    layers.Dense(1, activation='sigmoid')
])

modelo_resnet.summary()
```

```python

# Entrenamos ResNet50 durante 10 épocas.
# Por ahora mantenemos congeladas sus capas originales.

modelo_resnet.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
    loss='binary_crossentropy',
    metrics=['accuracy']
)

resultado_resnet = modelo_resnet.fit(
    imagenes_train,
    validation_data=imagenes_val,
    epochs=10
)
```

```python

# Revisamos cómo ha sido el entrenamiento de ResNet50.

plt.figure(figsize=(12, 4))

plt.subplot(1, 2, 1)
plt.plot(resultado_resnet.history['accuracy'],
         label='Entrenamiento')
plt.plot(resultado_resnet.history['val_accuracy'],
         label='Validación')
plt.title('Accuracy ResNet50')
plt.xlabel('Épocas')
plt.ylabel('Accuracy')
plt.legend()

plt.subplot(1, 2, 2)
plt.plot(resultado_resnet.history['loss'],
         label='Entrenamiento')
plt.plot(resultado_resnet.history['val_loss'],
         label='Validación')
plt.title('Loss ResNet50')
plt.xlabel('Épocas')
plt.ylabel('Loss')
plt.legend()

plt.tight_layout()
plt.show()
```

### Resultados del octavo experimento

ResNet50 obtuvo un accuracy del 99,76% en entrenamiento y del 99,17% en validación al terminar, con una pérdida de validación de 0,0394. Las gráficas muestran que el modelo aprende rápido (98,27% de validación en la primera época) y que los resultados de validación se mantienen altos durante casi todo el entrenamiento con una única bajada en la época 3 (97,29%).

Frente a MobileNetV2 congelada (98,64% y una pérdida de 0,0409), ResNet50 queda por encima apenas por unos pocos puntos de accuracy, que equivalen a unas siete imágenes de las 1.326 de validación. Por lo tanto no se observa una mejora clara por utilizar una red unas diez veces mayor en parámetros. Como en los experimentos anteriores, esa validación contiene imágenes muy parecidas a las de entrenamiento así que se tiene que validar si la diferencia se mantiene con las imágenes de prueba. Por esta razon conservaremos también esta versión de ResNet50 antes de realizar fine-tuning. Después podremos comparar ambos estados del modelo con las mismas imágenes de prueba.

## Revisión de coincidencias entre imágenes

Antes de comparar los nueve modelos quieremos comprobar un detalle del dataset: si algunas fotografías aparecen repetidas o son casi iguales entre entrenamiento, validación y prueba.

En una primera revisión muchas piezas se parecen bastant así que no basta con confiar en que los archivos estén en carpetas distintas. Si una fotografía de prueba es idéntica a otra que el modelo ya vio el resultado podría ser demasiado bueno. Buscaremos archivos exactamente iguales y luego revisaremos similitudes visuales. Esta parte no cambia los modelos entrenados sino que sirve más bien para interpretar mejor su evaluación.

```python

import os
import hashlib

# Revisamos si tenemos imágenes exactamente iguales
# entre las carpetas de entrenamiento y prueba.

def revisar_imagenes(ruta):
    datos = {}

    for carpeta, subcarpetas, archivos in os.walk(ruta):
        for archivo in archivos:
            if archivo.lower().endswith(('.jpg', '.jpeg', '.png')):

                ruta_imagen = os.path.join(carpeta, archivo)

                with open(ruta_imagen, 'rb') as imagen:
                    codigo = hashlib.sha256(imagen.read()).hexdigest()

                datos.setdefault(codigo, []).append(ruta_imagen)

    return datos


# Obtenemos los códigos de las imágenes.
datos_train = revisar_imagenes(ruta_train)
datos_test = revisar_imagenes(ruta_test)

# Buscamos si hay imágenes idénticas en ambas carpetas.
repetidas = set(datos_train) & set(datos_test)

print("Imágenes únicas en entrenamiento:", len(datos_train))
print("Imágenes únicas en prueba:", len(datos_test))
print("Archivos idénticos entre ambas carpetas:", len(repetidas))
```

```python

# Revisamos las imágenes repetidas para saber
# si pertenecen a la misma categoría.

mismas_clases = 0
distintas_clases = 0

for codigo in repetidas:

    for imagen_train in datos_train[codigo]:
        for imagen_test in datos_test[codigo]:

            clase_train = os.path.basename(
                os.path.dirname(imagen_train)
            )

            clase_test = os.path.basename(
                os.path.dirname(imagen_test)
            )

            if clase_train == clase_test:
                mismas_clases += 1
            else:
                distintas_clases += 1

print("Coincidencias con la misma clase:", mismas_clases)
print("Coincidencias con diferente clase:", distintas_clases)
```

```python

# Vamos a separar las imágenes de prueba que no
# aparecen repetidas en entrenamiento.

imagenes_limpias = []
imagenes_repetidas = []

for codigo, rutas in datos_test.items():

    if codigo in datos_train:
        imagenes_repetidas.extend(rutas)
    else:
        imagenes_limpias.extend(rutas)

print("Imágenes de prueba originales:", 715)
print("Imágenes repetidas:", len(imagenes_repetidas))
print("Imágenes disponibles para prueba:", len(imagenes_limpias))
```

Después de retirar esas coincidencias quedan 651 archivos de prueba. Este método solo detecta copias exactas, no fotografías casi idénticas con pequeñas diferencias, así que a continuación revisaremos también las similitudes visuales.

```python

import hashlib
import os
import tensorflow as tf

# Vamos a revisar las mismas imágenes que TensorFlow
# utilizó para entrenamiento y validación.

# Creamos las listas de archivos con la misma división.
archivos_train = tf.keras.utils.image_dataset_from_directory(
    ruta_train,
    validation_split=0.2,
    subset='training',
    seed=42,
    image_size=tamano,
    batch_size=lote,
    label_mode='binary',
    shuffle=True
).file_paths

archivos_val = tf.keras.utils.image_dataset_from_directory(
    ruta_train,
    validation_split=0.2,
    subset='validation',
    seed=42,
    image_size=tamano,
    batch_size=lote,
    label_mode='binary',
    shuffle=True
).file_paths

# Calculamos un código para cada imagen.
def codigo_imagen(ruta):
    with open(ruta, 'rb') as archivo:
        return hashlib.sha256(archivo.read()).hexdigest()

codigos_train = set(codigo_imagen(ruta) for ruta in archivos_train)
codigos_val = set(codigo_imagen(ruta) for ruta in archivos_val)

# Comprobamos si hay imágenes idénticas en ambos conjuntos.
repetidas_train_val = codigos_train & codigos_val

print("Imágenes de entrenamiento:", len(archivos_train))
print("Imágenes de validación:", len(archivos_val))
print("Imágenes idénticas entre ambos:", len(repetidas_train_val))
```

```python

from PIL import Image
import imagehash
import numpy as np

# Calculamos un código visual para cada imagen.
def codigo_visual(ruta):
    with Image.open(ruta) as imagen:
        return imagehash.phash(imagen.convert('RGB'))

# Calculamos los códigos de entrenamiento y validación.
visuales_train = [codigo_visual(ruta) for ruta in archivos_train]
visuales_val = [codigo_visual(ruta) for ruta in archivos_val]

# Buscamos imágenes muy parecidas.
# Una distancia pequeña indica mayor similitud.
numeros_train = np.array(
    [list(map(int, hash_imagen.hash.flatten()))
     for hash_imagen in visuales_train],
    dtype=np.uint8
)

numeros_val = np.array(
    [list(map(int, hash_imagen.hash.flatten()))
     for hash_imagen in visuales_val],
    dtype=np.uint8
)

parecidas = []

for i, codigo_val in enumerate(numeros_val):
    distancias = np.count_nonzero(
        numeros_train != codigo_val,
        axis=1
    )

    posicion = np.argmin(distancias)
    distancia = int(distancias[posicion])

    if distancia <= 5:
        parecidas.append((
            archivos_val[i],
            archivos_train[posicion],
            distancia
        ))

print("Imágenes de validación con alguna muy parecida en entrenamiento:",
      len(parecidas))
```

```python

# Revisamos si las imágenes parecidas pertenecen
# a la misma categoría o a categorías diferentes.

misma_clase = 0
diferente_clase = 0

for ruta_val, ruta_ent, distancia in parecidas:

    clase_val = os.path.basename(os.path.dirname(ruta_val))
    clase_ent = os.path.basename(os.path.dirname(ruta_ent))

    if clase_val == clase_ent:
        misma_clase += 1
    else:
        diferente_clase += 1

print("Imágenes parecidas de la misma clase:", misma_clase)
print("Imágenes parecidas de diferente clase:", diferente_clase)
print("Total de imágenes revisadas:", len(parecidas))
```

## Evaluación de los nueve modelos

Con los entrenamientos terminados y el subconjunto de 587 imágenes preparado, ya podemos hacer una comparación utilizando las mismas fotografías para todos los modelos.

En las etiquetas del dataset, `def_front` corresponde a 0 y `ok_front` a 1. Vamos a calcular accuracy, precision, recall y F1, tomando las piezas defectuosas como la clase de interés. También contaremos directamente cuántos defectos se detectan y cuántos se dejan pasar.

Esta evaluación servirá para comparar las nueve configuraciones dentro de las condiciones de la práctica. No elimina la limitación de las fotografías similares que encontramos anteriormente.

```python

import tensorflow as tf
import numpy as np
import os

# Preparamos las imágenes de prueba que hemos seleccionado.
# Mantenemos el tamaño de 128 x 128 píxeles.

mis_imagenes = []
mis_etiquetas = []

for ruta in imagenes_finales:

    imagen = tf.keras.utils.load_img(
        ruta,
        target_size=(128, 128)
    )

    imagen = tf.keras.utils.img_to_array(imagen)

    # TensorFlow utiliza 0 para defectuosas
    # y 1 para correctas.
    clase = os.path.basename(os.path.dirname(ruta))

    if clase == 'def_front':
        etiqueta = 0
    else:
        etiqueta = 1

    mis_imagenes.append(imagen)
    mis_etiquetas.append(etiqueta)

mis_imagenes = np.array(mis_imagenes)
mis_etiquetas = np.array(mis_etiquetas)

print("Forma de las imágenes:", mis_imagenes.shape)
print("Forma de las etiquetas:", mis_etiquetas.shape)
print("Piezas defectuosas:", np.sum(mis_etiquetas == 0))
print("Piezas correctas:", np.sum(mis_etiquetas == 1))
```

### Recuperación de los modelos antes del fine-tuning

Para evaluar MobileNetV2 y ResNet50 en sus dos versiones necesitamos cargar los modelos que guardamos antes de ajustar sus últimas capas. Los archivos se crearon durante los experimentos sexto y octavo.

```python
# Recuperamos la versión de MobileNetV2 anterior al fine-tuning.
modelo_mobile_antes = tf.keras.models.load_model(
    'mobilenet_sin_fine.keras', compile=False
)

# Reconstruimos la estructura de ResNet50 para recuperar sus pesos.
# No entrenamos otro modelo: cargamos los pesos guardados anteriormente.
base_resnet_antes = tf.keras.applications.ResNet50(
    weights=None, include_top=False, input_shape=(128, 128, 3)
)

modelo_resnet_antes = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(128, 128, 3)),
    tf.keras.layers.Lambda(tf.keras.applications.resnet50.preprocess_input),
    base_resnet_antes,
    tf.keras.layers.GlobalAveragePooling2D(),
    tf.keras.layers.Dense(64, activation='relu'),
    tf.keras.layers.Dense(1, activation='sigmoid')
])

modelo_resnet_antes.load_weights('resnet_sin_fine.keras')

# Comprobamos que la arquitectura coincide en tamaño.
assert modelo_resnet_antes.count_params() == modelo_resnet.count_params()
print("Modelos anteriores al fine-tuning preparados.")
```

```python

from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix

# Creamos una función para revisar los resultados
# de cada modelo con las mismas imágenes.

def evaluar_mi_modelo(nombre, modelo):

    predicciones = modelo.predict(
        mis_imagenes,
        verbose=0
    ).flatten()

    clases = (predicciones >= 0.5).astype(int)

    # Consideramos como clase positiva las piezas defectuosas (0).
    accuracy = accuracy_score(mis_etiquetas, clases)
    precision = precision_score(mis_etiquetas, clases, pos_label=0, zero_division=0)
    recall = recall_score(mis_etiquetas, clases, pos_label=0, zero_division=0)
    f1 = f1_score(mis_etiquetas, clases, pos_label=0, zero_division=0)

    matriz = confusion_matrix(
        mis_etiquetas,
        clases,
        labels=[0, 1]
    )

    # Fila 0: piezas defectuosas reales.
    defectos_detectados = matriz[0, 0]
    defectos_no_detectados = matriz[0, 1]

    print("\nModelo:", nombre)
    print("Accuracy:", round(accuracy * 100, 2), "%")
    print("Precision defectuosas:", round(precision * 100, 2), "%")
    print("Recall defectuosas:", round(recall * 100, 2), "%")
    print("F1 defectuosas:", round(f1 * 100, 2), "%")
    print("Defectos detectados:", defectos_detectados)
    print("Defectos no detectados:", defectos_no_detectados)

    return [nombre, accuracy, precision, recall, f1,
            defectos_detectados, defectos_no_detectados]
```

```python
# Evaluamos los nueve modelos en el mismo orden del trabajo.
# Esta lista se crea una sola vez para evitar resultados duplicados.
mis_modelos = [
    ("CNN inicial", modelo_cnn),
    ("CNN con Dropout", modelo_dropout),
    ("CNN con Batch Normalization", modelo_batch),
    ("CNN con Data Augmentation", modelo_aumento),
    ("CNN con regularización L2", modelo_l2),
    ("MobileNetV2 sin Fine-tuning", modelo_mobile_antes),
    ("MobileNetV2 con Fine-tuning", modelo_mobilenet),
    ("ResNet50 sin Fine-tuning", modelo_resnet_antes),
    ("ResNet50 con Fine-tuning", modelo_resnet)
]

resultados_prueba = []
for nombre, modelo in mis_modelos:
    resultados_prueba.append(evaluar_mi_modelo(nombre, modelo))
```

```python
import pandas as pd

# Mostramos los resultados de todos los modelos en una sola tabla.
tabla_final = pd.DataFrame(
    resultados_prueba,
    columns=[
        'Modelo', 'Accuracy', 'Precision defectuosas',
        'Recall defectuosas', 'F1 defectuosas',
        'Defectos detectados', 'Defectos no detectados'
    ]
)
for columna in ['Accuracy', 'Precision defectuosas',
                'Recall defectuosas', 'F1 defectuosas']:
    tabla_final[columna] = (tabla_final[columna] * 100).round(2)

display(tabla_final)
```

### ¿Qué cambió con el fine-tuning?

La mejora más clara la encontramos en MobileNetV2. Antes del fine-tuning tenía un accuracy del 91,31%: detectaba todos los defectos, pero rechazaba 51 piezas correctas. Después alcanzó el 99,49% con dos rechazos incorrectos y una pieza defectuosa que no detectó. Llama la atención que en validación las dos versiones apenas se distinguían (98,64% y 99,25%), mientras que en prueba la diferencia es enorme. Esto refuerza la idea de que la validación con imágenes muy parecidas a las de entrenamiento nos daba una imagen demasiado optimista. Tambien es importante tener presente que la versión con fine-tuning acumula veinte épocas de entrenamiento frente a diez, y que cada modelo se ha entrenado una sola vez.

En ResNet50 el resultado es diferente, el accuracy se mantuvo en 93,19% antes y después del ajuste, y las matrices de confusión dieron los mismos totales. Eso no significa que el modelo haya quedado igual. Como ResNet50 detecta las 422 piezas defectuosas en las dos versiones, esas diez imágenes son todas piezas correctas: cinco pasaron de ser rechazadas a ser aceptadas y cinco al revés. Es decir, el fine-tuning sí modificó las predicciones de ResNet50, pero los cambios se compensaron y no mejoraron sus métricas globales en esta prueba.

```python
# Comparamos las predicciones de ResNet50 antes y después.
pred_antes = modelo_resnet_antes.predict(mis_imagenes, verbose=0).flatten()
pred_despues = modelo_resnet.predict(mis_imagenes, verbose=0).flatten()
clases_antes = (pred_antes >= 0.5).astype(int)
clases_despues = (pred_despues >= 0.5).astype(int)

mejoraron = sum((clases_antes != mis_etiquetas) &
                (clases_despues == mis_etiquetas))
empeoraron = sum((clases_antes == mis_etiquetas) &
                 (clases_despues != mis_etiquetas))

print("Diferencia media de probabilidades:",
      float(np.mean(np.abs(pred_antes - pred_despues))))
print("Imágenes que cambiaron:", int(np.sum(clases_antes != clases_despues)))
print("Imágenes que mejoraron:", int(mejoraron))
print("Imágenes que empeoraron:", int(empeoraron))
```

### Matrices de confusión

Para terminar la comparación vamos a mostrar las nueve matrices con las mismas etiquetas y en el mismo orden de los experimentos. Así se ve de manera más directa dónde se equivoca cada modelo, especialmente cuando clasifica una pieza defectuosa como correcta.

```python
import matplotlib.pyplot as plt
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay

# Mostramos las nueve matrices en una sola figura.
fig, graficas = plt.subplots(3, 3, figsize=(15, 13))
for i, (nombre, modelo) in enumerate(mis_modelos):
    predicciones = modelo.predict(mis_imagenes, verbose=0).flatten()
    clases = (predicciones >= 0.5).astype(int)
    matriz = confusion_matrix(mis_etiquetas, clases, labels=[0, 1])
    ConfusionMatrixDisplay(
        confusion_matrix=matriz,
        display_labels=['Defectuosa', 'Correcta']
    ).plot(ax=graficas.flat[i], cmap='Blues', colorbar=False)
    graficas.flat[i].set_title(nombre, fontsize=10)

plt.tight_layout()
plt.show()
```
